```python
import os
import httpx
from dotenv import load_dotenv

load_dotenv()
```

```python
headers = {"Authorization": os.getenv("FLASH_RAG_API_KEY", "")}
url = "https://flash-rag.staging.dataesr.ovh"
url_dev = "http://localhost:8000"
```

Update

```python
# update flash rag documents
payload = {
    "reference": "all",  # all, eesr, ssmesr
    "task": "populate",  # all, load, extract, transform, populate
    "use_cache": False,
    "use_fetch": False,
    "force_download": False,
    "force_ocr": False,
    "db_reset": False,
    "db_override": True,
}
response = httpx.post(url + "/update", headers=headers, json=payload)
response.json() if response.status_code == 200 else response.status_code
```

Query

```python
# query flash rag documents
payload = {
    "query": "quel est le nombre detudiant en 2024",
    "use_reranker": True,
    "k": 10,
    "filters": {"reference": "ssmesr"},  # eesr, ssmesr
}
response = httpx.post(url_dev + "/query", headers=headers, json=payload, timeout=5000)
response
```
