```python
import os
import httpx
from dotenv import load_dotenv

load_dotenv()
```

```python
headers = {"Authorization": os.getenv("FLASH_RAG_API_KEY", "")}
url = "https://flash-rag.staging.dataesr.ovh"
url_dev = "http://localhost:8000"
```

Update

```python
# update flash rag documents
payload = {
    "source": "all", # all, eesr, ssmesr
    "task": "all", # all, load", extract, transform", populate"
    "use_cache": True,
    "db_reset": False,
    "db_override": True,
}
response = httpx.post(url + "/update", headers=headers, json=payload)
response
```
